```python
from abc import ABC, abstractmethod


class BuildingSystem(ABC):
    """Abstract Base Class enforcing the contract for all automated building systems."""

    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

    @abstractmethod
    def status(self):
        pass


class AirConditioningSystem(BuildingSystem):
    """Child class representing the Air Conditioning system."""

    def start(self):
        print("Air Conditioning: Compressor started, cooling auditorium to 21°C.")

    def stop(self):
        print("Air Conditioning: Compressor shut down.")

    def status(self):
        print("Air Conditioning Status: Active - Target Temperature Reached.")


class LightingSystem(BuildingSystem):
    """Child class representing the Auditorium Lighting system."""

    def start(self):
        print("Lighting System: Dimming main stage lights to 80% brightness.")

    def stop(self):
        print("Lighting System: All auditorium lights powered off.")

    def status(self):
        print("Lighting System Status: Active - Preset 'Lecture Mode'.")


class SecuritySystem(BuildingSystem):
    """Child class representing the Building Security system."""

    def start(self):
        print("Security System: Electronic locks engaged and motion sensors armed.")

    def stop(self):
        print("Security System: Main entrances unlocked, sensors set to standby.")

    def status(self):
        print("Security System Status: Active - All perimeters secure.")


class FireAlarmSystem(BuildingSystem):
    """
    Newly added child class inheriting from BuildingSystem.
    Must implement all abstract methods to strictly enforce the contract.
    """

    def start(self):
        print("Fire Alarm System: High-frequency siren activated & smoke vents open!")

    def stop(self):
        print("Fire Alarm System: Alarm silenced and reset to monitoring state.")

    def status(self):
        print("Fire Alarm System Status: Active - Optical smoke detectors operational.")


def main():
    # Instantiate objects of each subclass
    systems = [
        AirConditioningSystem(),
        LightingSystem(),
        SecuritySystem(),
        FireAlarmSystem()  # Integrated without altering the processing logic
    ]

    print("=== UMaT Auditorium Building Automation Control ===")

    # Polymorphic execution calling all required abstract methods
    print("\n--- Starting All Systems ---")
    for system in systems:
        system.start()

    print("\n--- Checking Systems Status ---")
    for system in systems:
        system.status()

    print("\n--- Stopping All Systems ---")
    for system in systems:
        system.stop()


if __name__ == "__main__":
    main()
```
